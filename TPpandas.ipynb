```python
import pandas as pd

emissions = pd.read_csv("https://koumoul.com/s/data-fair/api/v1/datasets/igt-pouvoir-de-rechauffement-global/convert")
emissions.head(20)
```

```python
newdf = emissions.drop(columns = ["Agriculture", "CO2 biomasse hors-total", "Déchets", "Energie", "Industrie hors-énergie", "Résidentiel", "Routier", "Tertiaire"])
newdf.head(2)
newdf = newdf.rename(columns ={"INSEE commune" : "Code_INSEE", "Autres transports" : "Transport", "Autres transports international" : "Transport international"})
```

```python
newdf.fillna(0, inplace = True)
newdf.head()

```

```python
newdf["dep"] = newdf["Code_INSEE"].astype(str).str[:2]
newdf["transports_total"] = newdf["Transport"] + newdf["Transport international"]
newdf
```
